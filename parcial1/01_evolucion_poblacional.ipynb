## Actividad 3 — Estadística descriptiva

Filtro: **Afghanistan**, período **2005–2023**.

```python
df_A = df[(df["Year"] >= 2005) & (df["Year"] <= 2023) & (df["Entity"] == "Afghanistan")]
df_A.head(19)
```

```python
df_A.mean(numeric_only=True)
```

```python
df_A.median(numeric_only=True)
```

```python
df_A.var(numeric_only=True)
```

```python
df_A.std(numeric_only=True)
```
